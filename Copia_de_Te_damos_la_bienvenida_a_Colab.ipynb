```
import math

# Clase Color
class Color:
    def __init__(self, color):
        self.__color = color

    def get_color(self):
        return self.__color

    def set_color(self, color):
        self.__color = color


# Clase FiguraGeometrica
class FiguraGeometrica:
    def __init__(self, ancho, alto):
        self.__ancho = ancho
        self.__alto = alto

    def get_ancho(self):
        return self.__ancho

    def set_ancho(self, ancho):
        if ancho <= 0:
            raise ValueError("El ancho debe ser mayor a 0")
        self.__ancho = ancho

    def get_alto(self):
        return self.__alto

    def set_alto(self, alto):
        if alto <= 0:
            raise ValueError("El alto debe ser mayor a 0")
        self.__alto = alto

    def __str__(self):
        return f"Ancho: {self.__ancho}, Alto: {self.__alto}"


# Clase Rectangulo (herencia múltiple)
class Rectangulo(FiguraGeometrica, Color):
    def __init__(self, ancho, alto, color):
        FiguraGeometrica.__init__(self, ancho, alto)
        Color.__init__(self, color)

    def area(self):
        return self.get_ancho() * self.get_alto()

    def __str__(self):
        return (f"Rectángulo -> {super().__str__()}, "
                f"Color: {self.get_color()}, Área: {self.area()}")


# Clase Cuadrado
class Cuadrado(Rectangulo):
    def __init__(self, lado, color):
        super().__init__(lado, lado, color)

    def __str__(self):
        return (f"Cuadrado -> Lado: {self.get_ancho()}, "
                f"Color: {self.get_color()}, Área: {self.area()}")


# Clase Circunferencia
class Circunferencia(FiguraGeometrica, Color):
    def __init__(self, radio, color):
        super().__init__(radio, radio)
        Color.__init__(self, color)

    def area(self):
        return math.pi * self.get_ancho() ** 2

    def __str__(self):
        return (f"Circunferencia -> Radio: {self.get_ancho()}, "
                f"Color: {self.get_color()}, Área: {round(self.area(), 2)}")


# PRUEBA
if __name__ == "__main__":
    r = Rectangulo(4, 5, "rojo")
    c = Cuadrado(3, "azul")
    circ = Circunferencia(2.5, "verde")

    print(r)
    print(c)
    print(circ)
```

```
import math

# Clase Color
class Color:
    def __init__(self, color):
        self.color = color

    @property
    def color(self):
        return self.__color

    @color.setter
    def color(self, valor):
        if not valor:
            raise ValueError("El color no puede estar vacío")
        self.__color = valor


# Clase FiguraGeometrica
class FiguraGeometrica:
    def __init__(self, ancho, alto):
        self.ancho = ancho
        self.alto = alto

    @property
    def ancho(self):
        return self.__ancho

    @ancho.setter
    def ancho(self, valor):
        if valor <= 0:
            raise ValueError("El ancho debe ser mayor a 0")
        self.__ancho = valor

    @property
    def alto(self):
        return self.__alto

    @alto.setter
    def alto(self, valor):
        if valor <= 0:
            raise ValueError("El alto debe ser mayor a 0")
        self.__alto = valor

    def __str__(self):
        return f"Ancho: {self.ancho}, Alto: {self.alto}"


# Rectángulo
class Rectangulo(FiguraGeometrica, Color):
    def __init__(self, ancho, alto, color):
        FiguraGeometrica.__init__(self, ancho, alto)
        Color.__init__(self, color)

    def area(self):
        return self.ancho * self.alto

    def perimetro(self):
        return 2 * (self.ancho + self.alto)

    def __str__(self):
        return f"Rectángulo -> {super().__str__()}, Color: {self.color}"


# Cuadrado
class Cuadrado(Rectangulo):
    def __init__(self, lado, color):
        super().__init__(lado, lado, color)

    def __str__(self):
        return f"Cuadrado -> Lado: {self.ancho}, Color: {self.color}"


# Circunferencia
class Circunferencia(FiguraGeometrica, Color):
    def __init__(self, radio, color):
        super().__init__(radio, radio)
        Color.__init__(self, color)

    def area(self):
        return math.pi * self.ancho ** 2

    def perimetro(self):
        return 2 * math.pi * self.ancho

    def __str__(self):
        return f"Circunferencia -> Radio: {self.ancho}, Color: {self.color}"


# PRUEBA
if __name__ == "__main__":
    r = Rectangulo(4, 5, "rojo")
    c = Cuadrado(3, "azul")
    circ = Circunferencia(2.5, "verde")

    print(r)
    print("Área:", r.area(), "Perímetro:", r.perimetro())
    print(c)
    print("Área:", c.area(), "Perímetro:", c.perimetro())
    print(circ)
    print("Área:", round(circ.area(), 2), "Perímetro:", round(circ.perimetro(), 2))
```

```
class Persona:
    def __init__(self, nombre, genero, edad=None, direccion=None):
        self.__nombre = nombre
        self.__genero = genero
        self.__edad = None
        self.__direccion = None

        if edad is not None:
            self.edad = edad  # usa el setter
        if direccion is not None:
            self.direccion = direccion  # usa el setter

    # nombre
    @property
    def nombre(self):
        return self.__nombre

    @nombre.setter
    def nombre(self, valor):
        self.__nombre = valor

    # genero
    @property
    def genero(self):
        return self.__genero

    @genero.setter
    def genero(self, valor):
        self.__genero = valor

    # edad (opcional, pero >= 18 si se define)
    @property
    def edad(self):
        return self.__edad

    @edad.setter
    def edad(self, valor):
        if valor is not None and valor < 18:
            raise ValueError("La edad debe ser mayor o igual a 18")
        self.__edad = valor

    # direccion (opcional)
    @property
    def direccion(self):
        return self.__direccion

    @direccion.setter
    def direccion(self, valor):
        self.__direccion = valor

    # __str__ usando __dict__
    def __str__(self):
        return str(self.__dict__)


class Cliente:
    def __init__(self, idcliente, fecha_registro=None, vip=False):
        self.__idcliente = idcliente
        self.__fecha_registro = fecha_registro
        self.__vip = vip

    # idcliente
    @property
    def idcliente(self):
        return self.__idcliente

    @idcliente.setter
    def idcliente(self, valor):
        self.__idcliente = valor

    # fecha_registro (opcional)
    @property
    def fecha_registro(self):
        return self.__fecha_registro

    @fecha_registro.setter
    def fecha_registro(self, valor):
        self.__fecha_registro = valor

    # vip (opcional, booleano)
    @property
    def vip(self):
        return self.__vip

    @vip.setter
    def vip(self, valor):
        if not isinstance(valor, bool):
            raise ValueError("El valor de VIP debe ser True o False")
        self.__vip = valor

    # representación del objeto
    def __str__(self):
        return f"Cliente(idcliente={self.__idcliente}, fecha_registro={self.__fecha_registro}, vip={self.__vip})"


class Empleado(Persona):
    def __init__(self, nombre, genero, edad=None, direccion=None, sueldo=None):
        super().__init__(nombre, genero, edad, direccion)
        self.__sueldo = sueldo

    @property
    def sueldo(self):
        return self.__sueldo

    @sueldo.setter
    def sueldo(self, valor):
        self.__sueldo = valor

    def __str__(self):
        return (f"Empleado(Nombre={self.nombre}, Genero={self.genero}, Edad={self.edad}, "
                f"Direccion={self.direccion}, Sueldo={self.sueldo})")


class EmpleadoCliente(Empleado, Cliente):
    def __init__(self, nombre, genero, edad=None, direccion=None, sueldo=None, idcliente=None, fecha_registro=None, vip=False, fechaRegistro=None):
        Empleado.__init__(self, nombre=nombre, genero=genero, edad=edad, direccion=direccion, sueldo=sueldo)
        Cliente.__init__(self, idcliente=idcliente, fecha_registro=fecha_registro, vip=vip)
        self.__fechaRegistro = fechaRegistro

    def __str__(self):
        empleado_str = Empleado.__str__(self)
        cliente_str = Cliente.__str__(self)
        empleado_content = empleado_str.replace('Empleado(', '')[:-1]
        cliente_content = cliente_str.replace('Cliente(', '')[:-1]
        return f"EmpleadoCliente({empleado_content}, {cliente_content}, FechaRegistroAdicional={self.__fechaRegistro})"


if __name__ == "__main__":
    ce1 = EmpleadoCliente(nombre="Carlos", genero="Masculino", edad=25, direccion="Av. Siempre Viva",
                          idcliente=10,
                          vip=True,
                          sueldo=500,
                          fecha_registro="2023-07-01",
                          fechaRegistro="2023-07-01")
    print(ce1)
```
